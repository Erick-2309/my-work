```python
import numpy as np   #library to manipulate arrays
import matplotlib.pyplot as plt  #library to plot figures
import scipy.signal as sci  #signal processing library
from scipy.io import wavfile
# %matplotlib qt
```

# Exercise 1

```python
"""
Exercise 1:
Generate a sine wave (amplitude A=1, normalized frequency fo=0.09814, 
nbs=32768 samples) add a white Gaussian noise (np.random.randn(nbs)) 
so that the SNR will be 20 dB.
Plot the temporal view
Plot the psd over M= 1024 frequencies.
try different values of M=(32768, 1024, 256) your conclusions?
"""


nbs=32768 #number of samples for the simulation
A=1 # sine wave amplitude
fo=100.5/1024 #0.09814
snr=20 # SNR in dB
n=np.arange(nbs) # the time index
s=A*np.sin(2*np.pi*n*fo) #the sine wave
b=np.random.randn(nbs) # the white Gaussian noise (its power is 1, check by computing np.cov(b))
Ps=A**2/2 #theoretical sine wave power, check by computing np.cov(s)
Pb=Ps/(10**(snr/10)) #snr=10*np.log10(Ps/Pb)
x=s+np.sqrt(Pb)*b # the power of the signal np.sqrt(Pb)*b is Pb because the power of b is 1

plt.figure()
plt.subplot(211)
plt.plot(n,x)
plt.xlabel("time index (n)")
plt.ylabel("amplitude (xn))")
plt.subplot(212)
for M in (32768,1024,256):
    f,h=plt.psd(x,M,Fs=1,label=f"M={M}") #the greater M, the greater the frequency resolution but the larger estimation noise
plt.legend()
```
