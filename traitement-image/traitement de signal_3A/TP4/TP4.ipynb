```python
import numpy as np #library to manipulate arrays
import matplotlib.pyplot as plt #library to plot figures
import scipy.signal as sci #signal processing library
from scipy.io import wavfile
%matplotlib qt
```

```python
"""
Compute the impulse response h(n).
Compute the frequency response. What kind of filter is it ?
What can you say about the filter stability ?
By using the function z,p,k=scipy.signal.tf2zpk(b,a), 
plot the poles and the zeros in the complex plane. States the filter stability.
"""

sig2=np.loadtxt("TP2_sig1.txt")

plt.figure()
plt.plot(sig2)
plt.xlabel("temps normalisé")
plt.ylabel("amplitude")

plt.figure()
M=1024 #number of computed frequencies in psd
f,X=plt.psd(sig2,M,Fs=1) #use Fs=1 if you want normalized frequencies or Fs=100 for real frequencies in kHz
```

```python
coef=np.loadtxt("TP4.csv",delimiter=',')
b=coef[:,0] # non recursif coef
a=coef[:,1] # recursif coef
print(a)
print(b)

sig1=np.loadtxt("mysterySignal.txt")

y=sci.lfilter(b,a,sig1)
plt.figure()
plt.subplot(211)
plt.plot(y)

plt.subplot(212)
h,f=plt.psd(y,NFFT=1024,Fs=1)
h,f=plt.psd(sig1,NFFT=1024,Fs=1)
```
