```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# Fonction pour convertir une image RGB en niveaux de gris
def rgb2gray(rgb):
    r, g, b = rgb[:, :, 0], rgb[:, :, 1], rgb[:, :, 2] # Extraire les canaux R, G, B
    gray = 0.2989 * r + 0.5870 * g + 0.1140 * b
    return gray

# Fonction split pour séparer les canaux R, G, B
def split(image):
    # Convertir l'image en un tableau NumPy (si l'image est en mode RGB)
    img_array = np.array(image)
    
    # Extraire les canaux
    r = img_array[:, :, 0]  # Canal Rouge
    g = img_array[:, :, 1]  # Canal Vert
    b = img_array[:, :, 2]  # Canal Bleu
    
    return r, g, b

# Fonction merge pour recombiner les canaux R, G, B
def merge(r, g, b):
    # Créer une image vide pour chaque canal, en ajoutant des canaux vides pour les autres couleurs
    merged_image = np.stack((r, g, b), axis=-1)
    
    # Convertir le tableau NumPy en image PIL et retourner
    return Image.fromarray(merged_image)

# Exemple d'utilisation
# Charger une image
img = Image.open("CUB.png")

# Appliquer la fonction split()
r, g, b = split(img)

# Afficher les canaux séparés
plt.figure(figsize=(12, 4))

# Afficher le canal Rouge
plt.subplot(2, 3, 1)
plt.imshow(r, cmap='Reds')
plt.title('Canal Rouge')
plt.axis('off')

# Afficher le canal Vert
plt.subplot(2, 3, 2)
plt.imshow(g, cmap='Greens')
plt.title('Canal Vert')
plt.axis('off')

# Afficher le canal Bleu
plt.subplot(2, 3, 3)
plt.imshow(b, cmap='Blues')
plt.title('Canal Bleu')
plt.axis('off')


# Recréer l'image originale à partir des canaux séparés
reconstructed_img = merge(r, b, g)

# Afficher l'image reconstruite
plt.subplot(2, 3, 4)
plt.imshow(reconstructed_img)
plt.title('recombinaison')


#plt.tight_layout()
plt.show()
```

```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# Fonction split pour séparer les canaux R, G, B
def split(image):
    # Convertir l'image en un tableau NumPy (si l'image est en mode RGB)
    img_array = np.array(image)
    
    # Extraire les canaux
    r = img_array[:, :, 0]  # Canal Rouge
    g = img_array[:, :, 1]  # Canal Vert
    b = img_array[:, :, 2]  # Canal Bleu
    
    return r, g, b

# Fonction merge pour recombiner les canaux R, G, B
def merge(r, g, b):
    # Créer une image avec les trois canaux dans un ordre donné
    merged_image = np.stack((r, g, b), axis=-1)
    
    # Convertir le tableau NumPy en image PIL et retourner
    return Image.fromarray(merged_image)

# Charger l'image
img = Image.open("CUB.png")

# Appliquer la fonction split()
r, g, b = split(img)

# Affichage de l'image originale et des trois composantes
plt.figure()

# Afficher l'image originale
plt.subplot(2, 3, 1)
plt.imshow(img)
plt.title('Image Originale')
plt.axis('off')

# Afficher le canal Rouge
plt.subplot(2, 3, 2)
red_img = Image.merge("RGB", (Image.fromarray(r), Image.new('L', img.size), Image.new('L', img.size)))  # Garder que le rouge
plt.imshow(red_img)
plt.title('Composante Rouge')
plt.axis('off')

# Afficher le canal Vert
plt.subplot(2, 3, 3)
green_img = Image.merge("RGB", (Image.new('L', img.size), Image.fromarray(g), Image.new('L', img.size)))  # Garder que le vert
plt.imshow(green_img)
plt.title('Composante Verte')
plt.axis('off')

# Afficher le canal Bleu
plt.subplot(2, 3, 4)
blue_img = Image.merge("RGB", (Image.new('L', img.size), Image.new('L', img.size), Image.fromarray(b)))  # Garder que le bleu
plt.imshow(blue_img)
plt.title('Composante Bleue')
plt.axis('off')

# Recréer l'image en inversant les canaux (par exemple, B, R, G)
reconstructed_img = merge(b, r, g)

# Afficher l'image reconstruite avec les canaux inversés
plt.subplot(2, 3, 5)
plt.imshow(reconstructed_img)
plt.title('Reconstruction: B, R, G')
plt.axis('off')

# Recréer l'image avec les canaux dans l'ordre normal (R, G, B)
reconstructed_img_normal = merge(r, g, b)

# Afficher l'image reconstruite avec l'ordre normal
plt.subplot(2, 3, 6)
plt.imshow(reconstructed_img_normal)
plt.title('Reconstruction: R, G, B')
plt.axis('off')

# Affichage
plt.tight_layout()
plt.show()
```
