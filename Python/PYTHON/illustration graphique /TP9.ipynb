```python
"""
Exercice 4
Soient les fonctions suivantes :
a) y=x2–2x–15
b) y = (2x+7)/(3x-4)
c) y = racine (-x2+5)
d) y = racine (4x2+1)
1) Représentez en mode superposé les fonctions y = f(x) sur le domaine
 [-10,10[ (fixez les amplitudes min à -20 et max à +20).
2) Utilisez différentes couleurs et type de courbes.
3) A l’aide de la commande subplot, affichez toutes les fonctions dans une même fenêtre.
4) Ajouter une légende pour les axes et un titre pour chaque figure.
"""
import numpy as np
import matplotlib.pyplot as plt

# Domaine pour les fonctions
x = np.linspace(-10, 10, 1000)

# a) y = x^2 - 2x - 15
y1 = x**2 - 2*x - 15

# b) y = (2x + 7)/(3x - 4)
y2 = np.empty_like(x)
y2[:] = np.nan  # Initialiser avec NaN pour éviter la division par zéro
mask_b = (3*x - 4) != 0
y2[mask_b] = (2*x[mask_b] + 7) / (3*x[mask_b] - 4)

# c) y = √(-x^2 + 5)
y3 = np.empty_like(x)
y3[:] = np.nan # Initialiser avec NaN pour éviter d'avoir un nombre < 0
mask_c = (-x**2 + 5) >= 0
y3[mask_c] = np.sqrt(-x[mask_c]**2 + 5)

# d) y = √(4x^2 + 1)
y4 = np.sqrt(4*x**2 + 1)

# Créer les subplots
fig, axs = plt.subplots(2, 2, figsize=(12, 8))

# Fixer les amplitudes y de -20 à 20
for ax in axs.flatten():
    ax.set_ylim(-20, 20)
    ax.set_xlim(-10, 10)

# Tracer chaque fonction avec style différent
axs[0, 0].plot(x, y1, 'r-', label='y = x² - 2x - 15')
axs[0, 0].set_title("Fonction quadratique")
axs[0, 0].set_xlabel("x")
axs[0, 0].set_ylabel("y")
axs[0, 0].legend()

# Styles personnalisés : chaque courbe a un style unique ('r-', 'g--', 'b-.', 'm:')

axs[0, 1].plot(x, y2, 'g--', label='y = (2x + 7)/(3x - 4)')
axs[0, 1].set_title("Fonction rationnelle")
axs[0, 1].set_xlabel("x")
axs[0, 1].set_ylabel("y")
axs[0, 1].legend()

axs[1, 0].plot(x, y3, 'b-.', label='y = √(-x² + 5)')
axs[1, 0].set_title("Fonction racine (domaine limité)")
axs[1, 0].set_xlabel("x")
axs[1, 0].set_ylabel("y")
axs[1, 0].legend()

axs[1, 1].plot(x, y4, 'm:', label='y = √(4x² + 1)')
axs[1, 1].set_title("Fonction racine (toujours définie)")
axs[1, 1].set_xlabel("x")
axs[1, 1].set_ylabel("y")
axs[1, 1].legend()

# Ajuster l'espacement
plt.tight_layout()
plt.show()
```

```python
"""
Exercice 5
Soient les fonctions suivantes :
a) z=sin(x).cos(y)pourxε[0à2π[etyε[0à2π[
b) z = 100.sin(x).sin(y).exp(-x2 + x.y – y2) pour x ε [-2 à 2[ et y ε [-2 à 2[
c) z = exp(x/10)-1 pour x ε [0 à 20[ et y=x.sin(x
d) z =sinc(x).sinc(y)pourxε[-6à6[etyε[-6à6[
1) Représentez chacune des fonction z = f(x,y) sur leurs domaines respectifs (utilisez la fonction meshgrid , sic’est nécessaire) et avec un pas de 0.1 ou de 0.01(tout dépend de votre machine).
2) A l’aide de la souris, pivotez le graphe pour différentes vues
3) Utilisez différentes couleurs et type de courbes
4) Observez la transformation que la fonction meshgrid effectue et reproduisez là avec
une boucle for.
5) Re-testez l’affichage 3D en utilisant votre propre code qui remplace meshgrid.
"""
#changez color_map pour voir d'autres styles (viridis, plasma, cividis, coolwarm...)
""" La fonction np.meshgrid(x, y) prend deux vecteurs 1D (x et y) et crée deux tableaux 2D, X et Y, qui représentent toutes les paires possibles de coordonnées (x, y).

import numpy as np
x = [1, 2, 3]
y = [10, 20]
X, Y = np.meshgrid(x, y)

resultat:
X = [[1 2 3]     Y = [[10 10 10]
     [1 2 3]]         [20 20 20]]
"""



import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Fonction sinc définie en numpy
def sinc(x):
    return np.sinc(x / np.pi)

# Création d’un meshgrid custom (remplace np.meshgrid)
def custom_meshgrid(x, y):
    X = np.zeros((len(y), len(x))) # crée une matrice len(y) lignes × len(x) colonnes
    Y = np.zeros((len(y), len(x))) 
    for i in range(len(y)):
        for j in range(len(x)):
            X[i, j] = x[j]    # Remplit chaque ligne de X avec x
            Y[i, j] = y[i]    # Remplit chaque colonne de Y avec y
            
    return X, Y

# Fonction pour affichage 3D
def plot_surface(X, Y, Z, title, color_map='viridis', style='surface'):
    fig = plt.figure()
    ax = fig.add_subplot(111, projection='3d')
    if style == 'surface':
        surf = ax.plot_surface(X, Y, Z, cmap=color_map, edgecolor='k', linewidth=0.2)
    elif style == 'wireframe':
        ax.plot_wireframe(X, Y, Z, color='black')
    elif style == 'contour':
        ax.contour3D(X, Y, Z, 50, cmap=color_map)
    ax.set_title(title)
    plt.show()

# === a) z = sin(x) * cos(y) sur [0, 2π[ x [0, 2π[
x = np.arange(0, 2*np.pi, 0.01)
y = np.arange(0, 2*np.pi, 0.01)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)
plot_surface(X, Y, Z, 'a) z = sin(x) * cos(y)', color_map='coolwarm')

# === b) z = 100*sin(x)*sin(y)*exp(-x^2 + x*y - y^2)
x = np.arange(-2, 2, 0.01)
y = np.arange(-2, 2, 0.01)
X, Y = np.meshgrid(x, y)
Z = 100 * np.sin(X) * np.sin(Y) * np.exp(-X**2 + X*Y - Y**2)
plot_surface(X, Y, Z, 'b) Fonction gaussienne modifiée', color_map='plasma')

# === c) z = exp(x/10) - 1 et y = x*sin(x)
x = np.arange(0, 20, 0.01)
y = x * np.sin(x)
X, Y = custom_meshgrid(x, y)  # Pour s'assurer que les dimensions marchent ( Car y dépend de x )
#X, Y = np.meshgrid(x, y) 
Z = np.exp(X / 10) - 1
plot_surface(X, Y, Z, "c) z = exp(x/10)-1, y = x*sin(x)", color_map='inferno')

# === d) z = sinc(x)*sinc(y)
x = np.arange(-6, 6, 0.01)
y = np.arange(-6, 6, 0.01)
"""
X, Y = np.meshgrid(x, y) 
Z = sinc(X) * sinc(Y)
plot_surface(X,Y, Z, 'd) z = sinc(x)*sinc(y)', color_map='cividis', style='contour')

"""
X_custom, Y_custom = custom_meshgrid(x, y)  
Z = sinc(X_custom) * sinc(Y_custom)
plot_surface(X_custom, Y_custom, Z, 'd) z = sinc(x)*sinc(y)', color_map='cividis', style='contour')

```
