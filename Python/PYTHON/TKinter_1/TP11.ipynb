#
1. Suppression de global x1, y1, x2, y2 dans la fonction drawline
Ce qui se passe : Aucune ligne n'est tracée.
Pourquoi : Sans l'instruction global, Python considère que les variables x1, y1, x2, y2 utilisées dans la fonction sont locales. Si elles ne sont pas définies dans la fonction, une erreur se produit ou les variables sont simplement ignorées. La fonction ne peut donc pas accéder aux valeurs des variables globales définies en dehors de la fonction.

2. Passage de x1, y1, x2, y2 comme paramètres de la fonction drawline
Ce qui se passe : Le programme fonctionne correctement.
Pourquoi : En passant les variables comme paramètres, la fonction reçoit explicitement les valeurs nécessaires pour tracer la ligne. Cela rend la fonction indépendante des variables globales.

3. Définition de x1, y1, x2, y2 = 10, 390, 390, 10 dans la fonction
Ce qui se passe : Une ligne est tracée, mais elle utilise toujours les mêmes coordonnées (10, 390, 390, 10), quelle que soit la situation.
Pourquoi : Les variables sont définies localement dans la fonction. Elles ne dépendent plus des valeurs globales ou des paramètres passés. Cela rend la fonction rigide et incapable de tracer des lignes avec d'autres coordonnées.
